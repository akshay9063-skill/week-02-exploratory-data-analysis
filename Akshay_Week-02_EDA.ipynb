Objective: explore the Iris dataset using statistics, visualisations, and correlation analysis.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('cleaned_iris_dataset.csv')
df.head()
```

## Dataset overview

```python
print('Shape:', df.shape)
print('\nData types:\n', df.dtypes)
print('\nMissing values:\n', df.isna().sum())
print('\nDuplicate rows:', df.duplicated().sum())
print('\nSpecies counts:\n', df['species'].value_counts())
```

## Descriptive statistics

```python
numeric_cols = ['sepal_length_cm','sepal_width_cm','petal_length_cm','petal_width_cm']
df[numeric_cols].describe().round(2)
```

## Histogram

```python
for species in sorted(df.species.unique()):
    plt.hist(df.loc[df.species == species, 'petal_length_cm'], bins=12, alpha=0.55, label=species)
plt.title('Petal Length Distribution by Species')
plt.xlabel('Petal length (cm)'); plt.ylabel('Frequency'); plt.legend(); plt.tight_layout(); plt.show()
```

## Scatter plot

```python
for species in sorted(df.species.unique()):
    d = df[df.species == species]
    plt.scatter(d.petal_length_cm, d.petal_width_cm, label=species, alpha=0.8)
plt.title('Petal Length vs Petal Width'); plt.xlabel('Petal length (cm)'); plt.ylabel('Petal width (cm)'); plt.legend(); plt.tight_layout(); plt.show()
```

## Box plot

```python
species_order = sorted(df.species.unique())
plt.boxplot([df.loc[df.species == s, 'sepal_length_cm'] for s in species_order], labels=species_order)
plt.title('Sepal Length by Species'); plt.xlabel('Species'); plt.ylabel('Sepal length (cm)'); plt.tight_layout(); plt.show()
```

## Correlation heatmap

```python
corr = df[numeric_cols].corr()
display(corr.round(2))
plt.imshow(corr, vmin=-1, vmax=1); plt.colorbar(label='Pearson correlation')
labels = [c.replace('_',' ') for c in numeric_cols]
plt.xticks(range(4), labels, rotation=35, ha='right'); plt.yticks(range(4), labels)
for i in range(4):
    for j in range(4): plt.text(j, i, f'{corr.iloc[i,j]:.2f}', ha='center', va='center')
plt.title('Correlation Heatmap'); plt.tight_layout(); plt.show()
```

## Species-wise means and conclusions

```python
df.groupby('species')[numeric_cols].mean().round(2)
```

### Findings
- Petal length and petal width have a strong positive correlation (approximately 0.96).
- Setosa has smaller average petal measurements than the other species.
- Sepal measurements overlap more than petal measurements.
- Correlation does not prove causation. The dataset is small and educational, so avoid overgeneralising.
